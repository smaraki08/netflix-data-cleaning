Identify and handle missing values column by column


```python
netflix_data.isnull().sum()
```

```python
netflix_data.isnull().sum().plot(kind="bar")
plt.title("Missing data of each column")
```

```python
netflix_data.fillna({'show_id':"Missing data"	,'type':"Missing data"	,'title':"Missing data"	,'director':"Missing data",	'cast':"Missing data",	'country':"Missing data"	,'date_added':0	,'release_year':0	,'rating':0	,'duration':0	,'listed_in':"Missing data"	,'description':"Missing data"},inplace=True)
netflix_data.head(10)
```

Fix mixed-type columns (e.g., duration stored as "90 min")

```python
netflix_data['duration'].str.split(" ").str[0].astype(float)
```

Parse date columns into proper datetime objects

```python
netflix_data.columns = netflix_data.columns.str.strip()

netflix_data['date_added'] = pd.to_datetime(netflix_data['date_added'],errors='coerce')
```
